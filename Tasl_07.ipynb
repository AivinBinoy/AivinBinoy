```python
import pandas as pd
import numpy as np
```

```python
data = {
    'Age': [25, 30, np.nan, 45, 50, np.nan, 35],
    'Blood_Pressure': [120, np.nan, 130, 140, np.nan, 135, 128],
    'Cholesterol': [200, 220, np.nan, 240, 230, np.nan, 210],
    'Gender': ['M', 'F', 'F', np.nan, 'M', 'F', np.nan]
}

df = pd.DataFrame(data)
df
```

## 🔍 Dataset Inspection
Checking structure and missing values

```python
df.info()
df.describe()
df.isnull().sum()
```

## 📌 Mean Imputation
Replacing missing values using mean (for numerical columns)

```python
df_mean = df.copy()

df_mean['Age'].fillna(df_mean['Age'].mean(), inplace=True)
df_mean['Blood_Pressure'].fillna(df_mean['Blood_Pressure'].mean(), inplace=True)
df_mean['Cholesterol'].fillna(df_mean['Cholesterol'].mean(), inplace=True)

df_mean
df_mean.describe()
```

## 📌 Median Imputation
Replacing missing values using median

```python
df_median = df.copy()

df_median['Age'].fillna(df_median['Age'].median(), inplace=True)
df_median['Blood_Pressure'].fillna(df_median['Blood_Pressure'].median(), inplace=True)
df_median['Cholesterol'].fillna(df_median['Cholesterol'].median(), inplace=True)

df_median
df_median.describe()
```

## 📌 Mode Imputation
Replacing missing values in categorical column

```python
df_mode = df.copy()

df_mode['Gender'].fillna(df_mode['Gender'].mode()[0], inplace=True)

df_mode
df_mode['Gender'].value_counts()
```

## 📊 Comparison of Imputation Methods

```python
print("Mean Imputation:\n", df_mean.describe())
print("\nMedian Imputation:\n", df_median.describe())
```

## 🔍 Observations

- Mean imputation changes the average values slightly and is sensitive to outliers.
- Median imputation is more stable and not affected by extreme values.
- Mode imputation fills categorical data using the most frequent value.

### 📌 Key Differences:
- Mean shifts distribution if outliers exist
- Median keeps distribution balanced
- Mode maintains category frequency

## ✅ When to Use Each Method

- **Mean**: When data is normally distributed and no outliers
- **Median**: When data has outliers or skewed distribution
- **Mode**: For categorical data

## 🧾 Conclusion

All three imputation techniques were applied and compared.  
Median imputation proved more robust for numerical data with missing values, while mode was effective for categorical data.  
Proper selection of imputation method is important to maintain data integrity.
