```python
import pandas as pd
from sklearn.preprocessing import StandardScaler, MinMaxScaler

# Load dataset
df = pd.read_csv("fitness_activity_100+_preprocess.csv")

# View data
print(df.head())
```

```python
# Select numeric columns
numeric_cols = df.select_dtypes(include=['int64', 'float64'])

print(numeric_cols.head())
```

# **Standard Scaling (Z-score Normalization)**

👉 Formula:

Z=
(X−μ)/σ
	​

```python
scaler_standard = StandardScaler()

standard_scaled = scaler_standard.fit_transform(numeric_cols)

standard_df = pd.DataFrame(standard_scaled, columns=numeric_cols.columns)

print(standard_df.head())
```

```python
standard_df.describe()
```

```python
standard_df.describe()
```

### Standard Scaling Observations

After applying Standard Scaling, the data is transformed such that its mean is very close to 0 and its standard deviation is very close to 1. This is evident from the `describe()` output for `standard_df`.

- **Mean**: The `mean` for each column is approximately `1.199e-16`, `-3.552e-17`, and `-3.108e-17` respectively, which are all extremely close to zero.
- **Standard Deviation**: The `std` for each column is `1.005038e+00`, which is very close to 1.
- **Range**: The range of values is not fixed; it depends on the original data's distribution. Outliers will still be present and can significantly affect the scaled values, pushing them further from the mean.

# **Min-Max Scaling (Normalization)**

👉 Formula:

X'=
(X-Xmin)/
X max - X min

```python
scaler_minmax = MinMaxScaler()

minmax_scaled = scaler_minmax.fit_transform(numeric_cols)

# Convert to DataFrame
minmax_df = pd.DataFrame(minmax_scaled, columns=numeric_cols.columns)

print(minmax_df.head())
```

```python
minmax_df.describe()
```

### Min-Max Scaling Observations

Min-Max Scaling transforms the data to a fixed range, typically between 0 and 1. This is confirmed by the `describe()` output for `minmax_df`.

- **Minimum**: The `min` for all columns is `0.000000`.
- **Maximum**: The `max` for all columns is `1.000000`.
- **Range**: All scaled values now lie strictly between 0 and 1.
- **Impact of Outliers**: While the values are squeezed into a specific range, Min-Max scaling is sensitive to outliers. A single large outlier can compress the majority of the data into a very small range, reducing the effectiveness of the scaling for the bulk of the data points.

```python
print("Standard Scaling Summary:\n", standard_df.describe())
print("\nMin-Max Scaling Summary:\n", minmax_df.describe())
```

### Comparison and Key Differences

Both Standard Scaling and Min-Max Scaling are common normalization techniques, but they differ in their approach and impact on the data:

1.  **Range of Values**:
    *   **Standard Scaling**: Transforms data to have a mean of 0 and a standard deviation of 1. The resulting values do not have a fixed range.
    *   **Min-Max Scaling**: Transforms data to a fixed range, typically 0 to 1. The minimum value becomes 0, and the maximum value becomes 1.

2.  **Sensitivity to Outliers**:
    *   **Standard Scaling**: Less affected by outliers than Min-Max scaling. Outliers will still exist but do not compress the rest of the data into a smaller range as much.
    *   **Min-Max Scaling**: Highly sensitive to outliers. Extreme values can disproportionately affect the scaling, leading to most of the data points being compressed into a narrow interval, which might reduce the effectiveness of distinguishing between them.

3.  **Use Cases**:
    *   **Standard Scaling**: Often preferred when the algorithm assumes a Gaussian distribution or when algorithms like SVM, Logistic Regression, or Neural Networks are used, where features with larger variances might dominate.
    *   **Min-Max Scaling**: Useful when the data needs to be within a specific range (e.g., for image processing, neural networks with sigmoid activation functions, or algorithms that don't assume any distribution of the data).

In this analysis, we can clearly see the distinct characteristics in the `describe()` outputs: Standard Scaling provides values centered around zero, while Min-Max Scaling provides values bounded between 0 and 1.
